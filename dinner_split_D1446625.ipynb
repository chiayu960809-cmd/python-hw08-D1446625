## 作業說明

本作業使用 Python 製作聚餐費用試算器，透過不同資料型別與運算方式，計算聚餐的餐費、服務費、折價金額、總費用及每人平均分攤金額，並判斷是否超過預算。

## 測試一 : 基本資料

```python
# 測試一：基本資料

# 1. 建立原始資料
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"

service_rate = 0.10
discount = 100
budget_per_person = 300

# 2. 整理文字並轉換資料型別
event_clean = event_name.strip().lower()
meal_clean = float(meal_amount.replace("NT$", "").replace(",", "").strip())
people_count = int(people)

# 3. 計算費用
service_fee = meal_clean * service_rate
total = meal_clean + service_fee - discount
per_person = total / people_count

# 4. 判斷是否超過預算
over_budget = per_person > budget_per_person

# 5. 輸出費用摘要
print(f"聚餐名稱：{event_clean}")
print(f"原始餐費：{meal_clean:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total:.2f} 元")
print(f"分攤人數：{people_count} 人")
print(f"每人費用：{per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{over_budget}")
```
